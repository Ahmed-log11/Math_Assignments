# Mission 1: Understanding the Matrix

**Manual:**

$$X = \begin{bmatrix} 120 & 3 & 10 \\ 150 & 4 & 8 \\ 180 & 4 & 6 \\ 210 & 5 & 4 \end{bmatrix}$$

- Shape = **(4, 3)** → **4 rows**, **3 columns**
- Each **row** = one house (H1, H2, H3, H4)
- Each **column** = one feature (Area, Bedrooms, Age)

```python
import numpy as np

X = np.array([
    [120, 3, 10],
    [150, 4, 8],
    [180, 4, 6],
    [210, 5, 4]
])

print(X)
print("Shape:", X.shape)
print("Rows:", X.shape[0])
print("Columns:", X.shape[1])
```

# Mission 2: Matrix Operations

**Manual:** (Age column = [10, 8, 6, 4], Area column = [120, 150, 180, 210])

- Age + 2 = [12, 10, 8, 6]
- Age − 1 = [9, 7, 5, 3]
- Area × 1.1 = [132, 165, 198, 231]
- Transpose: rows become columns, shape (4, 3) → (3, 4)

$$X^T = \begin{bmatrix} 120 & 150 & 180 & 210 \\ 3 & 4 & 4 & 5 \\ 10 & 8 & 6 & 4 \end{bmatrix}$$

```python
X1 = X.copy()
X1[:, 2] = X1[:, 2] + 2
print("Age + 2:\n", X1)

X2 = X.copy()
X2[:, 2] = X2[:, 2] - 1
print("Age - 1:\n", X2)

X3 = X.astype(float)
X3[:, 0] = X3[:, 0] * 1.1
print("Area * 1.1:\n", X3)

print("Transpose:\n", X.T)
print("Transpose shape:", X.T.shape)
```

**AI Insight:**
- **Change values:** addition, subtraction and multiplication (the numbers change, the shape stays (4, 3)).
- **Change structure:** transpose (the shape changes from (4, 3) to (3, 4)).
- **Why transpose is useful:** it lets us switch between rows and columns so the shapes match for matrix multiplication (for example $X^T X$ in linear regression).

# Mission 4: Detecting Redundant Features

**Manual:** Rooms = Bedrooms × 2, so the Rooms column gives **no new information**.
It is a copy of another column (linearly dependent).
The matrix has 4 columns but only **3 independent columns**, so **rank = 3**.

```python
X_new = np.array([
    [120, 3, 10, 6],
    [150, 4, 8, 8],
    [180, 4, 6, 8],
    [210, 5, 4, 10]
])

print(X_new)
print("Shape:", X_new.shape)
print("Rank:", np.linalg.matrix_rank(X_new))
```

**Findings:** The new matrix has **4 columns but rank 3**. The rank is less than the number of columns, which proves one feature is **redundant**: Rooms is just Bedrooms × 2.

**AI Insight:** Redundant features are harmful because they add no new information but make the model **bigger and slower**, can cause **overfitting**, and make the matrix **singular** (det = 0), so methods that need an inverse (like linear regression) become unstable.

# Mission 5: Determinant, Identity Matrix & Inverse

**Manual:** $\det\begin{bmatrix} a & b \\ c & d \end{bmatrix} = ad - bc$

- det(A) = (2)(2) − (1)(4) = 4 − 4 = **0**
- det(B) = (2)(2) − (1)(3) = 4 − 3 = **1**

```python
A = np.array([[2, 1],
              [4, 2]])

B = np.array([[2, 1],
              [3, 2]])

I = np.array([[1, 0],
              [0, 1]])

print("det(A) =", round(np.linalg.det(A), 2))
print("det(B) =", round(np.linalg.det(B), 2))
```

**Verify A × I = A and B × I = B** (multiplying by the identity matrix changes nothing):

```python
print("A @ I =\n", A @ I)
print("Equal to A?", np.array_equal(A @ I, A))

print("B @ I =\n", B @ I)
print("Equal to B?", np.array_equal(B @ I, B))
```

**Which matrix is invertible?** **B**, because det(B) = 1 ≠ 0.

**Manual inverse of B:**

$$B^{-1} = \frac{1}{\det B}\begin{bmatrix} d & -b \\ -c & a \end{bmatrix} = \frac{1}{1}\begin{bmatrix} 2 & -1 \\ -3 & 2 \end{bmatrix} = \begin{bmatrix} 2 & -1 \\ -3 & 2 \end{bmatrix}$$

```python
B_inv = np.linalg.inv(B)
print("Inverse of B:\n", B_inv)

print("Check B @ B_inv =\n", B @ B_inv)
```

**Why A has no inverse:** det(A) = 0. The second row of A is 2 × the first row ([4, 2] = 2 × [2, 1]), so the rows are dependent. The inverse formula divides by the determinant, and we **cannot divide by 0**.

**AI Insight:**
- **det = 0** means the matrix is **singular**: its rows/columns are dependent and it squashes space into a lower dimension (information is lost).
- **A singular matrix can't be inverted** because the lost information can't be recovered, and the formula would divide by 0.
- **The inverse is important in ML** because it is used to solve for the best weights, for example in linear regression: $w = (X^T X)^{-1} X^T y$.

# Mission 6: Solving Linear Systems

$$2x + y = 7 \qquad x - y = 2$$

**Matrix representation** $A\,x = b$:

$$\begin{bmatrix} 2 & 1 \\ 1 & -1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 7 \\ 2 \end{bmatrix}$$

**Manual solution** (add the two equations):
- (2x + y) + (x − y) = 7 + 2 → 3x = 9 → **x = 3**
- 3 − y = 2 → **y = 1**

```python
A_sys = np.array([[2, 1],
                  [1, -1]])
b = np.array([7, 2])

solution = np.linalg.solve(A_sys, b)
print("x =", solution[0])
print("y =", solution[1])
```

**Comparison:** Manual solution: x = 3, y = 1. NumPy solution: x = 3, y = 1. **Both solutions are the same.**

**AI Insight:** Systems of equations are important in Machine Learning because training a model means finding the weights that satisfy many equations at once (one equation per data sample). Linear regression is solved as a system of equations ($X^T X\,w = X^T y$).
