# Mission 1: Discovering Eigenvectors

$$A = \begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix} \qquad v_1 = \begin{bmatrix} 1 \\ 1 \end{bmatrix} \qquad v_2 = \begin{bmatrix} 1 \\ -1 \end{bmatrix}$$

**Manual:**

$$A v_1 = \begin{bmatrix} 2(1) + 1(1) \\ 1(1) + 2(1) \end{bmatrix} = \begin{bmatrix} 3 \\ 3 \end{bmatrix} = 3 \begin{bmatrix} 1 \\ 1 \end{bmatrix} = 3\,v_1$$

$$A v_2 = \begin{bmatrix} 2(1) + 1(-1) \\ 1(1) + 2(-1) \end{bmatrix} = \begin{bmatrix} 1 \\ -1 \end{bmatrix} = 1\,v_2$$

```python
import numpy as np

A = np.array([[2, 1],
              [1, 2]])

v1 = np.array([1, 1])
v2 = np.array([1, -1])

print("A @ v1 =", A @ v1)
print("A @ v2 =", A @ v2)
```

**Comparing directions:**
- $v_1 = [1, 1]$ → $[3, 3]$: **same direction**, only stretched 3 times.
- $v_2 = [1, -1]$ → $[1, -1]$: **same direction**, same length (multiplied by 1).

**Both vectors stay on the same direction.**

**Why are they called eigenvectors?** Because the matrix only **scales** them ($A v = \lambda v$) and does not change their direction. An eigenvector is a special vector whose direction is kept by the transformation.

For comparison, a normal vector like $[1, 0]$ changes direction:

```python
u = np.array([1, 0])
print("A @ u =", A @ u)
```

$[1, 0]$ became $[2, 1]$, which points in a **different direction**, so $[1, 0]$ is **not** an eigenvector.

**Insight:** A matrix stretches space differently in different directions, so most vectors get pushed and **rotated** to a new direction. Only a few special directions (the eigenvectors) are stretched along themselves, so they **stay on the same line**.

# Mission 2: Finding Eigenvalues

**Characteristic equation:** $\det(A - \lambda I) = 0$

$$A - \lambda I = \begin{bmatrix} 2-\lambda & 1 \\ 1 & 2-\lambda \end{bmatrix}$$

$$\det(A - \lambda I) = (2-\lambda)(2-\lambda) - (1)(1) = 0$$

$$\lambda^2 - 4\lambda + 4 - 1 = 0 \;\;\Rightarrow\;\; \lambda^2 - 4\lambda + 3 = 0$$

$$(\lambda - 3)(\lambda - 1) = 0$$

**Eigenvalues:** $\lambda_1 = 3$, $\lambda_2 = 1$

```python
values = np.linalg.eigvals(A)
print("Eigenvalues:", values.real)
```

NumPy gives **3 and 1**, the same as the manual solution. ✅

**AI Insight:** An eigenvalue tells us **how much the eigenvector is stretched or shrunk** by the transformation (and whether it is flipped if negative). Here, the direction $[1, 1]$ is stretched 3 times, and the direction $[1, -1]$ stays the same size.

# Mission 3: Finding Eigenvectors

Solve $(A - \lambda I)\,v = 0$ for each eigenvalue.

**For λ = 3:**

$$\begin{bmatrix} -1 & 1 \\ 1 & -1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = 0 \;\Rightarrow\; -x + y = 0 \;\Rightarrow\; y = x \;\Rightarrow\; v_1 = \begin{bmatrix} 1 \\ 1 \end{bmatrix}$$

**For λ = 1:**

$$\begin{bmatrix} 1 & 1 \\ 1 & 1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = 0 \;\Rightarrow\; x + y = 0 \;\Rightarrow\; y = -x \;\Rightarrow\; v_2 = \begin{bmatrix} 1 \\ -1 \end{bmatrix}$$

**Normalized (length = 1)**, divide by $\|v\| = \sqrt{1^2 + 1^2} = \sqrt{2}$:

$$v_1 = \begin{bmatrix} 0.707 \\ 0.707 \end{bmatrix} \qquad v_2 = \begin{bmatrix} 0.707 \\ -0.707 \end{bmatrix}$$

```python
values, vectors = np.linalg.eig(A)

print("Eigenvalues:", values.real)
print("Eigenvectors (columns):\n", vectors.real)
```

```python
v1_norm = v1 / np.linalg.norm(v1)
v2_norm = v2 / np.linalg.norm(v2)

print("v1 normalized:", v1_norm)
print("v2 normalized:", v2_norm)
```

**Comparison:** NumPy returns normalized eigenvectors as **columns**. The first column $[0.707, 0.707]$ matches $v_1$. The second column $[-0.707, 0.707]$ is $-1 \times [0.707, -0.707]$, which is the **same eigenvector** (an eigenvector can be multiplied by any non-zero number, including −1, and it stays on the same line). ✅

**AI Insight:** Each eigenvalue is a different stretching amount, and each stretching happens along its **own direction**. Solving $(A - \lambda I)v = 0$ gives a different solution for each λ, so every eigenvalue has its own eigenvector.

# Mission 4: Geometric Interpretation

$$A v = \lambda v$$

| Eigenvalue | Interpretation |
|---|---|
| **λ = 3** | The eigenvector is **stretched 3 times longer** and keeps the **same direction**. |
| **λ = 0.5** | The eigenvector is **shrunk to half** its length and keeps the **same direction**. |
| **λ = −2** | The eigenvector is **stretched 2 times** and its **direction is flipped** (points the opposite way). |

```python
v = np.array([1, 1])

print("lambda = 3   :", 3 * v)
print("lambda = 0.5 :", 0.5 * v)
print("lambda = -2  :", -2 * v)
```

**AI Insight:**
- The **size** of λ controls **scaling**: |λ| > 1 stretches, |λ| < 1 shrinks, |λ| = 1 keeps the same length.
- The **sign** of λ controls **direction**: positive keeps the direction, negative flips it.

# Mission 5: Matrix Diagonalization

**P** = eigenvectors as columns, **D** = eigenvalues on the diagonal:

$$P = \begin{bmatrix} 1 & 1 \\ 1 & -1 \end{bmatrix} \qquad D = \begin{bmatrix} 3 & 0 \\ 0 & 1 \end{bmatrix}$$

**Manual inverse of P:** $\det(P) = (1)(-1) - (1)(1) = -2$

$$P^{-1} = \frac{1}{-2}\begin{bmatrix} -1 & -1 \\ -1 & 1 \end{bmatrix} = \begin{bmatrix} 0.5 & 0.5 \\ 0.5 & -0.5 \end{bmatrix}$$

**Check:** $P D P^{-1} = A$

```python
P = np.array([[1, 1],
              [1, -1]])

D = np.array([[3, 0],
              [0, 1]])

P_inv = np.linalg.inv(P)
print("P inverse:\n", P_inv)

result = P @ D @ P_inv
print("P @ D @ P_inv =\n", result)
print("Equal to A?", np.allclose(result, A))
```

$P D P^{-1}$ gives back $A$. ✅

**Why are diagonal matrices easier?** A diagonal matrix has zeros everywhere except the diagonal, so:
- Multiplying by it just **scales each coordinate** separately.
- Its **powers** are easy: $D^n$ = each diagonal number to the power n (for example $D^{10} = \text{diag}(3^{10}, 1^{10})$), so $A^n = P D^n P^{-1}$.
- Its **determinant** is the product of the diagonal, and its **inverse** is 1 ÷ each diagonal number.

**AI Insight:** Diagonalization breaks a complex transformation into simple **independent scalings along the eigenvector directions**. In Machine Learning it is the basis of **PCA**, makes repeated matrix calculations fast, and helps analyze how models and algorithms behave over many steps.

# Mission 6: Eigenvalues in AI

**1. Why is PCA used?**
PCA (Principal Component Analysis) is used to **reduce the number of features** while keeping most of the important information. This makes models faster, reduces noise and overfitting, and allows us to visualize data in 2D or 3D.

**2. Why are eigenvalues important in PCA?**
PCA finds the eigenvectors and eigenvalues of the data's **covariance matrix**. Each eigenvector is a new direction (a principal component), and its **eigenvalue tells how much variance (information)** the data has in that direction.

**3. Why do we sort eigenvalues from largest to smallest?**
To **rank the directions by importance**. The largest eigenvalue is the direction with the most information, so sorting lets us easily pick the most important components first.

**4. Which eigenvectors should be selected?**
The eigenvectors with the **largest eigenvalues** (the top k components), usually enough of them to keep a high percentage of the total variance, for example 90–95%.

**5. What happens if we remove the eigenvector with the largest eigenvalue?**
We would lose the **most important information** in the data (the direction with the most variance). The model would perform much worse.

**Small example:** if the covariance matrix were our matrix A, its eigenvalues are 3 and 1:

```python
values = np.linalg.eigvals(A).real
values = np.sort(values)[::-1]

print("Sorted eigenvalues:", values)
print("Information kept by each:", values / values.sum() * 100, "%")
```

The first direction keeps **75%** of the information, so keeping only it reduces 2 features to 1 while keeping 75% of the variance.

**AI Insight:** Eigenvalues measure how much information each direction holds. By **keeping only the eigenvectors with the largest eigenvalues** and dropping the ones with small eigenvalues (which hold little information, mostly noise), we reduce the number of dimensions while **preserving most of the useful information**.
